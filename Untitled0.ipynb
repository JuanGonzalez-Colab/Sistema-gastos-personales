```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

# 1. CONECTAR BASE DE DATOS
conn = sqlite3.connect('gastos.db')
cursor = conn.cursor()

# 2. CREAR TABLA SI NO EXISTE
cursor.execute('''
CREATE TABLE IF NOT EXISTS gastos (
    id INTEGER PRIMARY KEY,
    fecha TEXT,
    categoria TEXT,
    descripcion TEXT,
    monto REAL
)
''')
conn.commit()

# 3. FUNCIONES CRUD
def agregar_gasto(categoria, descripcion, monto):
    fecha = datetime.now().strftime("%Y-%m-%d")
    cursor.execute("INSERT INTO gastos (fecha, categoria, descripcion, monto) VALUES (?, ?, ?, ?)",
                   (fecha, categoria, descripcion, monto))
    conn.commit()
    print(f"Gasto agregado: ${monto}")

def ver_gastos():
    df = pd.read_sql_query("SELECT * FROM gastos", conn)
    return df

def borrar_gasto(id_gasto):
    cursor.execute("DELETE FROM gastos WHERE id = ?", (id_gasto,))
    conn.commit()
    print("Gasto borrado")

def exportar_excel():
    df = ver_gastos()
    df.to_excel("reporte_gastos.xlsx", index=False)
    print("Reporte guardado como reporte_gastos.xlsx")

def mostrar_dashboard():
    df = ver_gastos()
    if df.empty:
        print("No hay gastos todavía")
        return

    gastos_por_cat = df.groupby('categoria')['monto'].sum()
    plt.figure(figsize=(8,5))
    gastos_por_cat.plot(kind='pie', autopct='%1.1f%%')
    plt.title('Gastos por Categoría')
    plt.ylabel('')
    plt.show()

# 4. PROBAR
agregar_gasto("Comida", "Almuerzo", 5000)
agregar_gasto("Transporte", "Subte", 800)
agregar_gasto("Comida", "Cena", 7000)

print("\n--- TUS GASTOS ---")
print(ver_gastos())

print("\n--- DASHBOARD ---")
mostrar_dashboard()

print("\n--- EXPORTAR ---")
exportar_excel()
```
