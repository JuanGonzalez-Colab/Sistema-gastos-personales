```python
import pandas as pd
import matplotlib.pyplot as plt

print("Cargando datos de ejemplo de Notebook Gamer...")

# DATOS DE EJEMPLO - COMO SI FUERAN DE ML HOY
data = {
    'Titulo': [
        'Notebook Gamer ASUS ROG Strix 16GB RTX 4060',
        'Notebook Gamer Lenovo Legion 5 32GB RTX 4070',
        'Notebook Gamer MSI Katana 16GB RTX 3050',
        'Notebook Gamer HP Victus 16GB GTX 1650',
        'Notebook Gamer Acer Nitro 5 16GB RTX 3060',
        'Notebook Gamer Dell G15 8GB RTX 3050',
        'Notebook Gamer Bangho Max L5 16GB RTX 4060',
        'Notebook Gamer Gigabyte G5 16GB RTX 3060'
    ],
    'Precio': [1850000, 2400000, 1200000, 950000, 1600000, 1100000, 1780000, 1550000],
    'Marca': ['ASUS', 'Lenovo', 'MSI', 'HP', 'Acer', 'Dell', 'Bangho', 'Gigabyte']
}

df = pd.DataFrame(data)
df.to_csv('precios_notebooks_final.csv', index=False)

print("Listo! Se guardaron 8 productos")
print(df)

# ANÁLISIS 1: PROMEDIO Y MÁS CARO
print(f"\nPrecio Promedio: ${df['Precio'].mean():,.0f}")
print(f"Notebook más cara: {df.loc[df['Precio'].idxmax()]['Titulo']}")

# ANÁLISIS 2: GRÁFICO POR MARCA
plt.figure(figsize=(10,5))
df.groupby('Marca')['Precio'].mean().sort_values().plot(kind='barh', color='orange')
plt.title('Precio Promedio por Marca - Notebook Gamer')
plt.xlabel('Precio en $')
plt.tight_layout()
plt.show()
```
